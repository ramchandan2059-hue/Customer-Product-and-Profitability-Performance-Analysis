Customer, Product, and Profitability Performance Analysis


Introduction:

In supply chain operations, success is often measured by speed and delivery performance, yet speed alone does not guarantee a healthy business. High sales do not always translate into high profit, and factors such as heavy discounting, shipping costs, and unprofitable customers or product categories can quietly erode margins. For a global logistics provider like APL Logistics (KWE Group), this creates a risk of making pricing and customer decisions that are revenue-focused but profit-blind.

This project analyzes detailed order and sales data to bring financial clarity to these decisions. It examines profitability across customers, products, categories, markets, and regions, and evaluates how discount levels affect profit margins. The analysis identifies high-value and low-value customers, high-revenue but low-margin products, loss-making categories, and the discount thresholds at which margins begin to collapse.
